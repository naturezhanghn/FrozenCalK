# FrozenCal-K playground
This notebook applies the released K-specific heads to frozen 12-dimensional feature vectors.

```python
import json
import numpy as np
from pathlib import Path
weights = json.loads(Path('weights.json').read_text())['methods']['FrozenCal-K']
features = weights['feature_names'][:12]
features
```

```python
def rank(rows, k):
    x = np.asarray(rows, dtype=float)
    z = (x - x.mean(0)) / np.maximum(x.std(0), 1e-4)
    score = np.concatenate([z, z], axis=1) @ np.asarray(weights['weights'][f'w{k}'])
    return np.argsort(-score), score

rows = np.array([[.72,.35,.76,.42,-.62,.09,.11,.87,.02,.01,.09,.72],
                  [.72,.35,.84,.42,-.62,.09,.11,.87,.02,.01,.09,.72]])
rank(rows, 2)
```
